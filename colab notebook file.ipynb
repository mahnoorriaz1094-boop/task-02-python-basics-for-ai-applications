```python
def recommend_service(requirement):

    if requirement.lower() == "chatbot":
        return "Chatbot Solutions"

    elif requirement.lower() == "data":
        return "Data Analytics"

    elif requirement.lower() == "ai":
        return "AI Development"

    else:
        return "Consultation Required"


print(recommend_service("chatbot"))
```
